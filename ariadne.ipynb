# Analyse results from Ariadne

```python
from pathlib import Path
import json
```

```python
from networkx.readwrite import json_graph
```

```python
from ariadne_roots.quantify import analyze
```

```python
fn = Path('json')/'RSPM_Ariadne_2025'/'H10'/'Hetero_D'/'p06_1.json'
```

```python
def analyze_with_ariadne(fn):
    with open(fn, mode="r") as h:
        data = json.load(h)
        graph = json_graph.adjacency_graph(data)
        results, front, randoms = analyze(graph)
        return results, front, randoms
```

```python
analyze_with_ariadne(fn)
```
